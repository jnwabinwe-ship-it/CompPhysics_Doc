```python
# Task 2 
import numpy as np
import matplotlib.pyplot as plt

#Setup Parameters
K = 0.05 #(min^-1)
T = 90.0 #(^o C)
T_env = 20.0 # (^o C)

dt = 2.0 # (min)
t = 0.0
t_final = 60.0

# Storing list
t_n = []
T_n = []

# Implementing function
def f(T):
    return -K * (T - T_env)

# RK4 equations
while t<= t_final:

    t_n.append(t) # Storing list
    T_n.append(T)

    k1 = f(T)
    k2 = f(T + (dt*0.5) * k1)
    k3 = f(T + (dt*0.5) * k2)
    k4 = f(T + dt * k3)

    T = T + (dt/6.0) * (k1 + 2*k2 + 2*k3 + k4) # temp with rk4

    t += dt # time with rk4

#Plotting
plt.figure(figsize=(8,5))
plt.plot(t_n, T_n, "o", label= "RK4")
plt.xlabel("Time(min)")
plt.ylabel("Temp(^o C)")
plt.title("Newton's Law of Cooling : RK4")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Task 3
import numpy as np
import matplotlib.pyplot as plt

#Setup Parameters
K = 0.05 #(min^-1)
T0 = 90.0 #(^o C)
T_env = 20.0 # (^o C)

dt = 2.0 # (min)
t = 0.0
t_final = 60.0

t_n = []
T_n = []

t = 0.0
T = T0

while t<= t_final :
    t_n.append(t)
    T_n.append(T)

    m = -K * (T - T_env) # slope
    T += m * dt
    t += dt

# RK4 
t = 0.0
T = T0

t_rk4 = []
T_rk4 = []

def f(T):
    return -K * (T - T_env)

# RK4 equations
while t<= t_final:

    t_rk4.append(t) # Storing list
    T_rk4.append(T)

    k1 = f(T)
    k2 = f(T + (dt*0.5) * k1)
    k3 = f(T + (dt*0.5) * k2)
    k4 = f(T + dt * k3)

    T = T + (dt/6.0) * (k1 + 2*k2 + 2*k3 + k4) # temp with rk4

    t += dt

# Analytic Solution
t_value = np.arange(0,t_final+dt, dt)
T_value = T_env + (T0 - T_env) * np.exp(-K * t_value)

#Plotting
plt.figure(figsize=(8,5))
plt.plot(t_n, T_n, "o", label="Euler")
plt.plot(t_rk4,T_rk4,"m.-", label="RK4")
plt.plot(t_value, T_value, "g-", label="Analytic Solution")
plt.xlabel("Time(min)")
plt.ylabel("Temp(^ oC)")
plt.title(" Newton's Law of Cooling : Euler vs RK4 vs Analytic")
plt.legend()
plt.grid()
plt.show()

# Q1 Response : the RK4 method would be closer to the exact solution; Euler is usally best for much smaller time steps.
#Q2 Response :  It looks off roughly by a degree Celsius or two.
```
