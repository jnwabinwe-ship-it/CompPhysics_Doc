```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. SETUP PARAMETERS
L = 1.0 #(m) 
g = 9.81 #(m/s^2)
dt = 0.01  #time step(s)
t_max = 15 # total time(s)

theta0 = 11 # initial angle
theta = np.deg2rad(theta0)
v_omega = 0.0 # initial velocity

# Array for Time
t = np.arange(0,t_max, dt)

# Lists to store the history (for plotting): Initialize these here! 
theta0_list = []
v_list = []


# 2. THE SIMULATION LOOP
for _ in t:
    
    
    # --- TODO: FILL IN THE PHYSICS HERE ---
    # 1. Calculate Acceleration (a = F_net / m)
    # Hint: Remember gra
    a = -(g/L) * np.sin(theta) # <--- REPLACE THIS

    # 2. Update Position (Euler Step)
    theta += v_omega * dt
      # <--- REPLACE THIS
    
    # 3. Update Velocity (Euler Step)
    v_omega += a * dt  # <--- REPLACE THIS
    

    
    # Store results
    theta0_list.append(theta)
    v_list.append(v_omega)




# 3. PLOT
#Make sure to add an x label and y label with units, and give it a title and legend. 
plt.figure(figsize=(10,5))
plt.plot(t,np.rad2deg(theta0_list))
plt.title("Pendulum Swing")
plt.xlabel("Time (s)")
plt.ylabel("Angle (Degrees/Theta)")
plt.grid(True)
plt.show()
```
